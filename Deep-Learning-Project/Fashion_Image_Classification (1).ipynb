# Fashion Image Classification Using ANN and CNN

**Dataset:** Fashion-MNIST (70,000 grayscale images, 10 classes). 54,000 training / 6,000 validation / 10,000 held-out test.

**Method:** Compare a dense ANN baseline with a convolutional neural network. Pixel values are scaled to [0,1]. Both models use Adam and sparse categorical cross-entropy. The held-out test set is evaluated only after training.

**Metrics:** Accuracy, macro precision, macro recall, macro F1, classification report and confusion matrix. Results depend on the run.

**Related work:** Xiao, Rasul & Vollgraf (2017), Fashion-MNIST: a Novel Image Dataset for Benchmarking Machine Learning Algorithms. https://arxiv.org/abs/1708.07747

```python
import os, random
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, classification_report, confusion_matrix, ConfusionMatrixDisplay

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
os.makedirs('results', exist_ok=True)
class_names = ['T-shirt/top','Trouser','Pullover','Dress','Coat','Sandal','Shirt','Sneaker','Bag','Ankle boot']

(x_train_all, y_train_all), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
# Fixed, shuffled validation split: 54,000 training / 6,000 validation / 10,000 test
rng = np.random.default_rng(SEED)
indices = rng.permutation(len(x_train_all))
val_idx, train_idx = indices[:6000], indices[6000:]
x_train, y_train = x_train_all[train_idx].astype('float32')/255.0, y_train_all[train_idx]
x_val, y_val = x_train_all[val_idx].astype('float32')/255.0, y_train_all[val_idx]
x_test = x_test.astype('float32')/255.0
print('Train:', x_train.shape, 'Validation:', x_val.shape, 'Test:', x_test.shape)

fig, axes = plt.subplots(2, 5, figsize=(10, 5))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_train[i], cmap='gray')
    ax.set_title(class_names[y_train[i]])
    ax.axis('off')
plt.tight_layout(); plt.savefig('results/sample_images.png', dpi=150); plt.show()

ann = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28,28)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
], name='ANN_baseline')
cnn = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(28,28,1)),
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu', padding='same'),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(10, activation='softmax')
], name='CNN_proposed')
for model in (ann, cnn):
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    model.summary()

EPOCHS = 10
BATCH_SIZE = 128
ann_history = ann.fit(x_train, y_train, validation_data=(x_val,y_val), epochs=EPOCHS, batch_size=BATCH_SIZE, verbose=1)
cnn_history = cnn.fit(x_train[...,np.newaxis], y_train, validation_data=(x_val[...,np.newaxis],y_val), epochs=EPOCHS, batch_size=BATCH_SIZE, verbose=1)

fig, axes = plt.subplots(1,2,figsize=(12,4))
for label, history in [('ANN',ann_history),('CNN',cnn_history)]:
    axes[0].plot(history.history['accuracy'], label=f'{label} train')
    axes[0].plot(history.history['val_accuracy'], linestyle='--', label=f'{label} validation')
    axes[1].plot(history.history['loss'], label=f'{label} train')
    axes[1].plot(history.history['val_loss'], linestyle='--', label=f'{label} validation')
axes[0].set_title('Accuracy by epoch'); axes[0].set_ylabel('Accuracy')
axes[1].set_title('Loss by epoch'); axes[1].set_ylabel('Loss')
for ax in axes:
    ax.set_xlabel('Epoch'); ax.legend(); ax.grid(alpha=0.2)
plt.tight_layout(); plt.savefig('results/training_curves.png',dpi=150); plt.show()

results = []
for label, model, test_inputs in [('ANN',ann,x_test),('CNN',cnn,x_test[...,np.newaxis])]:
    probabilities = model.predict(test_inputs, verbose=0)
    predictions = np.argmax(probabilities, axis=1)
    precision, recall, f1, _ = precision_recall_fscore_support(y_test,predictions,average='macro',zero_division=0)
    accuracy = accuracy_score(y_test,predictions)
    results.append((label,accuracy,precision,recall,f1))
    print(f'\n{label} TEST REPORT\n', classification_report(y_test,predictions,target_names=class_names,zero_division=0))
    cm = confusion_matrix(y_test,predictions)
    fig, ax = plt.subplots(figsize=(10,8))
    ConfusionMatrixDisplay(cm,display_labels=class_names).plot(ax=ax,xticks_rotation=45,cmap='Blues',colorbar=False)
    ax.set_title(f'{label} Test Confusion Matrix')
    plt.tight_layout(); plt.savefig(f'results/{label.lower()}_confusion_matrix.png',dpi=150); plt.show()

print('\nFINAL TEST COMPARISON (macro-averaged precision/recall/F1)')
print(f'{"Model":<8}{"Accuracy":>12}{"Precision":>12}{"Recall":>12}{"F1":>12}')
for name, acc, p, r, f1 in results:
    print(f'{name:<8}{acc:>12.4f}{p:>12.4f}{r:>12.4f}{f1:>12.4f}')
import csv
with open('results/model_comparison.csv','w',newline='') as f:
    writer=csv.writer(f); writer.writerow(['Model','Test Accuracy','Macro Precision','Macro Recall','Macro F1']); writer.writerows(results)

cnn_probs = cnn.predict(x_test[...,np.newaxis],verbose=0)
cnn_preds = np.argmax(cnn_probs,axis=1)
fig, axes = plt.subplots(3,4,figsize=(12,9))
for i, ax in enumerate(axes.flat):
    ax.imshow(x_test[i],cmap='gray')
    ax.set_title(f'Actual: {class_names[y_test[i]]}\nPredicted: {class_names[cnn_preds[i]]}',color=('green' if y_test[i]==cnn_preds[i] else 'red'),fontsize=9)
    ax.axis('off')
plt.tight_layout(); plt.savefig('results/cnn_predictions.png',dpi=150); plt.show()
cnn.save('fashion_cnn.keras')
print('Saved model: fashion_cnn.keras; graphs and metrics: results/')
```
